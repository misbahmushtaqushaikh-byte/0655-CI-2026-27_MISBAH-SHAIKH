```python
# %%
import matplotlib.pyplot as plt
```

```python
# %%
## create a line plot

plt.plot(df['Order Date'], df['customer_rating'])

plt.xlabel('Order Date')
plt.ylabel('Customer Rating')
plt.title("Basic Line Plot")

plt.xticks(rotation=45)
plt.show()
```

```python
# %%
## create a customized line plot

plt.plot(df['Order Date'], df['customer_rating'],
         color='blue',
         linestyle='--',
         marker='o',
         linewidth=3,
         markersize=9)

plt.xlabel('Order Date')
plt.ylabel('Customer Rating')
plt.title("Customized Line Plot")

plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
# %%
## multiple plots

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(df['Order Date'], df['customer_rating'], color='green')
plt.title("Plot 1 - Customer Rating")

plt.subplot(2,2,2)
plt.plot(df['Order Date'], df['quantity'], color='red')
plt.title("Plot 2 - Quantity")

plt.subplot(2,2,3)
plt.plot(df['Order Date'], df['price_inr'], color='blue')
plt.title("Plot 3 - Price")

plt.subplot(2,2,4)
plt.plot(df['Order Date'], df['Sales'], color='orange')
plt.title("Plot 4 - Sales")

plt.tight_layout()
plt.show()


```

```python
# %%
## bar plot

categories = df['city'].value_counts().head(5).index
values = df['city'].value_counts().head(5).values

## create a bar plot

plt.bar(categories, values, color='lightblue')

plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Swiggy City-wise Orders')

plt.xticks(rotation=45)
plt.show()
```

```python
# %%
## sample data

data = df['price_inr'].dropna()

## create a histogram

plt.hist(data, bins=5, color='lightblue', edgecolor='black')

plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Swiggy Price Distribution')

plt.show()
```

```python

## create a scatter plot

## sample data

x = df['price_inr']
y = df['customer_rating']

plt.scatter(x, y, color="blue", marker='x')

plt.xlabel('Price')
plt.ylabel('Customer Rating')
plt.title('Swiggy Price vs Rating')

plt.show()#
```

```python
# %%
## pie chart

city_counts = df['city'].value_counts()

labels = city_counts.head(4).index.tolist()
sizes = city_counts.head(4).values.tolist()

if len(city_counts) > 4:
    labels.append('Other')
    sizes.append(city_counts.iloc[4:].sum())

colors = ['gold', 'yellow', 'lightcoral', 'lightskyblue', 'lightgreen']

explode = (0.1, 0, 0, 0, 0)[:len(labels)]

## create a pie chart

plt.pie(sizes,
        explode=explode,
        labels=labels,
        colors=colors[:len(labels)],
        autopct="%1.1f%%")

plt.title("Swiggy City-wise Distribution")
plt.show()
```
